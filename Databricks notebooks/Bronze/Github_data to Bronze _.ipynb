# Github_data to Bronze

```python
device_mapping = spark.read.format('csv') \
.option('header' , True) \
.option('inferSchema', True)  \
.load('/Volumes/telecom_catalog/default/landing/Telecom_data/github_data/device_host_mapping_with_api.csv')


server_performance_data = spark.read.format('csv') \
.option('header' , True) \
.option('inferSchema', True)  \
.load('/Volumes/telecom_catalog/default/landing/Telecom_data/github_data/server_performance_logs (1).csv')
```

```python
device_mapping.display()
```

```python
(
    device_mapping.write.format("delta")
    .mode("overwrite")
    .save("/Volumes/telecom_catalog/default/bronze/Github_data/device_mapping")
)

(
    server_performance_data.write.format("delta")
    .mode("overwrite")
    .save("/Volumes/telecom_catalog/default/bronze/Github_data/Server_performance_logs")
)
```
