#Creating the Dimension Dim_device

## Overview
Creates **Dim_device** bridge dimension mapping device identifiers across three source systems.

**Source**: mapping_data (silver)
**Target**: telecom_catalog.gold_schema.dim_device

## Step 1: Create Dimension Table
Define dim_device with device_key (surrogate key) and three device identifiers: api_device_id, device_id, host_id.

```sql
CREATE OR REPLACE TABLE telecom_catalog.gold_schema.dim_device (
    device_key BIGINT GENERATED ALWAYS AS IDENTITY,
    api_device_id STRING NOT NULL,
    device_id STRING NOT NULL,
    host_id STRING NOT NULL
);
```

## Step 2: Load Device Mappings
Insert device identifier mappings from mapping_data. Auto-generated device_key provides surrogate key.

```sql
INSERT INTO telecom_catalog.gold_schema.dim_device (
    api_device_id,
    device_id,
    host_id
)
SELECT
    api_device_id,
    device_id,
    host_id
FROM telecom_catalog.silver_schema.mapping_data;
```

## Step 3: Validate Data
Verify device records and identifier mappings.

```sql
SELECT *
FROM telecom_catalog.gold_schema.dim_device
LIMIT 5;
```

---
## Summary
**Table**: `telecom_catalog.gold_schema.dim_device`

**Key Columns**:
* device_key (PK, auto-generated)
* api_device_id, device_id, host_id (device identifier mappings)

**Purpose**: Bridge table mapping device IDs across API, device logs, and server logs

**Load**: INSERT from mapping_data
